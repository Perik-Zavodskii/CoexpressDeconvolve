```python
from scipy.io import mmread, mmwrite
from scipy import sparse
import numpy as np
import gzip
from pathlib import Path

in_path  = Path("matrix.mtx.gz")
out_path = Path("matrix.mtx.gz")

# 1) Read (SciPy can handle .gz via file-like object)
with gzip.open(in_path, "rb") as f:
    M = mmread(f)          # sparse COO by default

# 2) Transpose (stays sparse); optionally switch to CSR for efficient row ops
Mt = M.transpose().tocsr()
Mt.sort_indices()          # nicer, canonical index order (optional)

# 3) Write back (gzipped)
with gzip.open(out_path, "wb") as f:
    # mmwrite will infer type/field; you can also set field='integer' if you know it's integer
    mmwrite(f, Mt)
```
