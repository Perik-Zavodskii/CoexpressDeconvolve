```python
import scanpy as sc
import tangram as tg
import pandas as pd
import numpy as np
import torch
import warnings
import h5py
from scipy.sparse import csr_matrix

warnings.filterwarnings('ignore')
device = 'cpu' # Or 'cuda' if a GPU is available
```

```python
# File Paths

sc_path = "sc_reference_full.h5ad" 
hvg_path_synth = "fixed_hvgs.csv"
hvg_path_real = "fixed_hvgs_real.csv"
gt_counts_path = "ground_truth_counts.csv"

synth_sp_path = "/Users/romanperikzavodskii/Desktop/4. Ery/CoDeconv/Breast Cancer/test_bench/Synthetic Visium/filtered_feature_bc_matrix.h5"
real_sp_path = '/Users/romanperikzavodskii/Desktop/4. Ery/CoDeconv/Breast Cancer/Visium/Breast Cancer/filtered_feature_bc_matrix.h5'


# Load & Process Ground Truth

print("Loading and converting ground truth counts to proportions...")
gt_counts = pd.read_csv(gt_counts_path, index_col='Spot')

# Exclude 'Total_Cells' to only divide the specific cell type columns
cell_types = [col for col in gt_counts.columns if col != 'Total_Cells']

# Convert counts to proportions
gt_props = gt_counts[cell_types].div(gt_counts['Total_Cells'], axis=0)
gt_props = gt_props.fillna(0) # In case Total_Cells was 0 for any spot (prevents NaNs)

gt_props.to_csv("ground_truth_proportions.csv")
print("Saved ground truth proportions to 'ground_truth_proportions.csv'.\n")


# Load Single-Cell Reference

print("Loading Single-Cell Reference (Whole Transcriptome)...")
ad_sc = sc.read_h5ad(sc_path)

# Ensure CellType is a categorical variable
ad_sc.obs['CellType'] = ad_sc.obs['CellType'].astype('category')

# Load Curated HVGs separately
hvgs_synth = pd.read_csv(hvg_path_synth, header=None)[0].tolist()
hvgs_real = pd.read_csv(hvg_path_real, header=None)[0].tolist()

print(f"Loaded {ad_sc.n_obs} cells and {ad_sc.n_vars} total genes.")
print(f"Loaded {len(hvgs_synth)} curated HVGs for synthetic spatial mapping.")
print(f"Loaded {len(hvgs_real)} curated HVGs for real spatial mapping.\n")
```

```python
# Tangram Pipeline Helper

def robust_read_10x_h5(filename):
    print(f"Manually loading H5 file: {filename}")
    with h5py.File(filename, 'r') as f:
        g = f['matrix']
        
        # Load the Sparse Matrix
        X = csr_matrix(
            (g['data'][:], g['indices'][:], g['indptr'][:]), 
            shape=g['shape'][:][::-1]
        )
        
        # Load Barcodes
        obs = pd.DataFrame(index=g['barcodes'][:].astype(str))
        
        # Load Features
        features = g['features']
        var = pd.DataFrame({
            'gene_ids': features['id'][:].astype(str),
            'feature_types': features['feature_type'][:].astype(str),
            'genome': features['genome'][:].astype(str)
        }, index=features['name'][:].astype(str))
        
        adata = sc.AnnData(X=X, obs=obs, var=var)
        adata.var_names_make_unique()
        return adata

def run_tangram_pipeline(ad_sc, sp_path, hvgs, prefix="tangram_synth", generate_props=True):
    print(f"Starting Pipeline for: {prefix}")

    try:
        ad_sp = sc.read_10x_h5(sp_path)
    except ValueError:
        ad_sp = robust_read_10x_h5(sp_path)
    
    ad_sp.var_names_make_unique()
    
    print("Aligning single-cell and spatial data on curated HVGs...")
    tg.pp_adatas(ad_sc, ad_sp, genes=hvgs)

    # Map Cells to Space
    print("Mapping cells to space (mode='cells')... This may take a few minutes.")
    ad_map = tg.map_cells_to_space(
        ad_sc, 
        ad_sp, 
        mode='cells', 
        density_prior='rna_count_based',
        device=device
    )
    
    # Extract Reconstructed Whole Transcriptome GEX
    print("Projecting FULL spatial gene expression...")
    ad_ge = tg.project_genes(adata_map=ad_map, adata_sc=ad_sc)
    
    # Extract the expression matrix (Spots x Genes) and save
    gex_df = pd.DataFrame(ad_ge.X, index=ad_ge.obs_names, columns=ad_ge.var_names)
    gex_df.to_csv(f"{prefix}_wt_gex.csv")
    print(f"Saved: {prefix}_wt_gex.csv")
    
    # Extract Cell Proportions
    if generate_props:
        print("Extracting cell proportions...")
        cell_types_ref = ad_sc.obs['CellType'].values
        
        # Create mapping DataFrame: Rows = CellTypes, Cols = Spots
        map_df = pd.DataFrame(ad_map.X, index=cell_types_ref, columns=ad_sp.obs_names)
        
        # Aggregate (sum) mapping probabilities by CellType
        spot_cell_probs = map_df.groupby(map_df.index).sum()
        
        # Transpose so Rows = Spots, Cols = CellTypes
        spot_cell_probs = spot_cell_probs.T
        
        # Normalize so each spot row sums to 1.0 (converting probability density to proportions)
        spot_props = spot_cell_probs.div(spot_cell_probs.sum(axis=1), axis=0)
        
        # Fill any empty spots with 0 to prevent NaNs
        spot_props = spot_props.fillna(0)
        
        spot_props.to_csv(f"{prefix}_props.csv")
        print(f"Saved: {prefix}_props.csv\n")
        
    return ad_map, ad_ge
```

```python
# 1. Run on Synthetic Data

ad_map_synth, ad_ge_synth = run_tangram_pipeline(
    ad_sc=ad_sc, 
    sp_path=synth_sp_path, 
    hvgs=hvgs_synth,               # Using synthetic HVGs
    prefix="tangram_synth", 
    generate_props=True
)
```

```python
# 2. Run on Real Data

ad_map_real, ad_ge_real = run_tangram_pipeline(
    ad_sc=ad_sc, 
    sp_path=real_sp_path, 
    hvgs=hvgs_real,                # Using real HVGs
    prefix="tangram_real", 
    generate_props=True
)

print("\nTangram benchmark runs complete!")
```
