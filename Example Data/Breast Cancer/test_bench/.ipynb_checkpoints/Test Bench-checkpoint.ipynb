```python
import os
import json
import numpy as np
import pandas as pd
from scipy.optimize import minimize_scalar
from sklearn.metrics import mean_squared_error

# Import step 1 from your existing pipeline
from codeconv import step1_acquisition_and_anchoring

# --- 1. Configuration & Loading ---
VISIUM_DIR = "synthetic_visium" # Path to your synthetic data output
CONFIG_PATH = "codeconv_config.json"
SPECIES = "Homo sapiens"
MIN_UMI = 300 # Same threshold used in your standard run

# Load the synthetic Visium data using your existing pipeline
slices = step1_acquisition_and_anchoring(VISIUM_DIR)
slice_name = list(slices.keys())[0]
sd = slices[slice_name]

# Load the ground truth counts
gt_df = pd.read_csv(os.path.join(VISIUM_DIR, "ground_truth_counts.csv"), index_col="Spot")

# Align ground truth with the filtered barcodes from Step 1
true_cells = gt_df.loc[sd.barcodes, "True_Cells"].values

# Load Housekeeping reference from config
with open(CONFIG_PATH, 'r') as f:
    cfg = json.load(f)
hk_reference = cfg["species_profiles"][SPECIES]['hk_profiles']

# --- 2. Precompute Static Variables for Speed ---
common_hk = [g for g in hk_reference.keys() if g in sd.gene_names]
hk_indices = [sd.gene_names.index(g) for g in common_hk]
ref_values_log = np.array([hk_reference[g] for g in common_hk])

# Calculate Spot-level HK signal (this doesn't change with the factor)
hk_counts_raw = sd.counts[:, hk_indices].toarray()
safe_total = sd.total_umi.copy()
safe_total[safe_total == 0] = 1
normalized_hk_log = np.log1p((hk_counts_raw / safe_total[:, np.newaxis]) * 10000)
spot_hk_log_means = np.mean(normalized_hk_log, axis=1)

standard_anchor_log_mean = np.mean(ref_values_log)
spot_signal_linear = np.expm1(spot_hk_log_means)

# Mask for spots that pass the UMI threshold
valid_mask = sd.total_umi >= MIN_UMI

# --- 3. Define the Objective Function ---
def calculate_rmse(factor):
    """Core mathematical logic of Step 2, returning RMSE against ground truth."""
    # Apply the tuning knob
    adjusted_standard_log_mean = standard_anchor_log_mean + np.log(factor)
    standard_signal_linear = np.expm1(adjusted_standard_log_mean)
    
    if standard_signal_linear < 0.001:
        standard_signal_linear = 0.001
        
    # Predict cell counts
    raw_n_cells = spot_signal_linear / standard_signal_linear
    predicted_cells = np.round(raw_n_cells).astype(int)
    
    # Filter out low quality spots (predict 0)
    predicted_cells[~valid_mask] = 0
    
    # Calculate RMSE strictly on valid spots to avoid background skew
    rmse = np.sqrt(mean_squared_error(true_cells[valid_mask], predicted_cells[valid_mask]))
    return rmse

# --- 4. Run the Optimizer ---
print("Optimizing anchor_mean_factor to minimize RMSE...")

# Bounded scalar optimization (looking for a multiplier between 0.1 and 10.0)
res = minimize_scalar(calculate_rmse, bounds=(0.1, 10.0), method='bounded')

optimal_factor = res.x
best_rmse = res.fun

print(f"Optimization Complete!")
print(f"Optimal anchor_mean_factor: {optimal_factor:.4f}")
print(f"Lowest RMSE achieved:       {best_rmse:.4f}")

# --- 5. Verify the Results ---
# Run one last time with the optimal factor to see the mean comparison
adjusted_standard_log_mean = standard_anchor_log_mean + np.log(optimal_factor)
standard_signal_linear = np.expm1(adjusted_standard_log_mean)
predicted_cells = np.round(spot_signal_linear / standard_signal_linear).astype(int)
predicted_cells[~valid_mask] = 0

print("\n--- Summary on Valid Spots ---")
print(f"Mean True Cells:      {np.mean(true_cells[valid_mask]):.2f}")
print(f"Mean Predicted Cells: {np.mean(predicted_cells[valid_mask]):.2f}")
```
