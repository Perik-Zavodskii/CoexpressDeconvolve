```python
import os
import json
import numpy as np
import pandas as pd
from scipy.optimize import minimize_scalar
from sklearn.metrics import mean_squared_error

# Import step 1 from your existing pipeline
from codeconv import step1_acquisition_and_anchoring

# --- 1. Configuration & Loading ---
VISIUM_DIR = "synthetic_visium" 
CONFIG_PATH = "codeconv_config.json"
SPECIES = "hs" 
MIN_UMI = 300 

# Unpack the items
counts, gene_names, barcodes, coords, total_umi, cfg_extra = step1_acquisition_and_anchoring(VISIUM_DIR)

# Load the ground truth counts
gt_df = pd.read_csv(os.path.join(VISIUM_DIR, "ground_truth_counts.csv"), index_col="Spot")

# Align ground truth with the filtered barcodes
true_cells = gt_df.loc[barcodes, "True_Cells"].values

# Load Housekeeping reference
with open(CONFIG_PATH, 'r') as f:
    cfg = json.load(f)
hk_reference = cfg["species_profiles"][SPECIES]['hk_profiles']

# --- 2. Precompute Static Variables for Speed ---
common_hk = [g for g in hk_reference.keys() if g in gene_names]
hk_indices = [gene_names.index(g) for g in common_hk]
ref_values_log = np.array([hk_reference[g] for g in common_hk])

hk_counts_raw = counts[:, hk_indices].toarray()
safe_total = total_umi.copy()
safe_total[safe_total == 0] = 1
normalized_hk_log = np.log1p((hk_counts_raw / safe_total[:, np.newaxis]) * 10000)
spot_hk_log_means = np.mean(normalized_hk_log, axis=1)

standard_anchor_log_mean = np.mean(ref_values_log)
spot_signal_linear = np.expm1(spot_hk_log_means)

valid_mask = total_umi >= MIN_UMI

# --- DIAGNOSTICS ---
print("--- DIAGNOSTICS ---")
mean_spot_signal = np.mean(spot_signal_linear[valid_mask])
base_ref_signal = np.expm1(standard_anchor_log_mean)
print(f"Mean Synthetic Spot HK Signal: {mean_spot_signal:.4f}")
print(f"Base Reference HK Signal:      {base_ref_signal:.4f}")

# Calculate Analytical Ideal Factor based on True Cells
target_signal = np.mean(spot_signal_linear[valid_mask] / true_cells[valid_mask])
ideal_factor = np.exp(np.log1p(target_signal) - standard_anchor_log_mean)
print(f"Analytically Ideal Factor:     {ideal_factor:.4f}\n")


# --- 3. CONTINUOUS OBJECTIVE FUNCTION ---
def calculate_rmse_smooth(factor):
    adjusted_standard_log_mean = standard_anchor_log_mean + np.log(factor)
    standard_signal_linear = np.expm1(adjusted_standard_log_mean)
    
    if standard_signal_linear < 0.001:
        standard_signal_linear = 0.001
        
    # FIX: No rounding here! Keep it continuous so the optimizer has a smooth gradient.
    raw_n_cells = spot_signal_linear / standard_signal_linear
    
    pred = raw_n_cells.copy()
    pred[~valid_mask] = 0
    
    return np.sqrt(mean_squared_error(true_cells[valid_mask], pred[valid_mask]))

# --- 4. Run the Optimizer ---
print("Optimizing anchor_mean_factor (Smooth Objective)...")
res = minimize_scalar(calculate_rmse_smooth, bounds=(0.0001, 10000.0), method='bounded')

optimal_factor = res.x

# --- 5. Verify the Final Discrete Results ---
# Run one last time with the optimal factor to see the DISCRETE mean comparison
adjusted_standard_log_mean = standard_anchor_log_mean + np.log(optimal_factor)
standard_signal_linear = np.expm1(adjusted_standard_log_mean)

# Now we apply the rounding logic exactly like the actual pipeline
final_predicted_cells = np.round(spot_signal_linear / standard_signal_linear).astype(int)
final_predicted_cells[~valid_mask] = 0

final_rmse = np.sqrt(mean_squared_error(true_cells[valid_mask], final_predicted_cells[valid_mask]))

print(f"\n--- OPTIMIZATION COMPLETE ---")
print(f"Optimal anchor_mean_factor: {optimal_factor:.4f}")
print(f"Final Discrete RMSE:        {final_rmse:.4f}")

print("\n--- Summary on Valid Spots ---")
print(f"Mean True Cells:      {np.mean(true_cells[valid_mask]):.2f}")
print(f"Mean Predicted Cells: {np.mean(final_predicted_cells[valid_mask]):.2f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr, spearmanr
import pandas as pd
import numpy as np
import matplotlib.lines as mlines

# Set publication-ready style
sns.set_style("whitegrid")
sns.set_context("paper", font_scale=1.2)

# 1. Create a DataFrame from the variables already in your memory
plot_df = pd.DataFrame({
    'True_Cells': true_cells[valid_mask],
    'Predicted_Cells': final_predicted_cells[valid_mask]
})

# 2. Calculate the final statistics
r_val, _ = pearsonr(plot_df['True_Cells'], plot_df['Predicted_Cells'])
rho_val, _ = spearmanr(plot_df['True_Cells'], plot_df['Predicted_Cells'])
rmse_val = 2.5369 # Passing the known RMSE from your output

plt.figure(figsize=(8, 6))

# 3. Plot Violins (Ground Truth vs. CoexpressDeconvolve Predictions)
ax = sns.violinplot(
    data=plot_df, 
    x='True_Cells', 
    y='Predicted_Cells', 
    inner=None, 
    palette="rocket", 
    density_norm="width",
    hue='True_Cells',
    legend=False
)

x_min = plot_df['True_Cells'].min()

# 4. Overlay Red Regression Line (Aligning x-axis categorical bins)
sns.regplot(
    data=plot_df, 
    x=plot_df['True_Cells'] - x_min, 
    y='Predicted_Cells', 
    scatter=False, 
    color='red', 
    line_kws={'linewidth': 2.5},
    ax=ax
)

# 5. Overlay 1:1 Perfect Prediction Line
x_vals = np.sort(plot_df['True_Cells'].unique())
plt.plot(
    x_vals - x_min, x_vals, 
    color='gray', linestyle='--', linewidth=2, zorder=4
)

# 6. Create a bulletproof custom legend
reg_line = mlines.Line2D([], [], color='red', linewidth=2.5, 
                         label=f'Regression Trend\nPearson $r$ = {r_val:.3f}\nSpearman $\\rho$ = {rho_val:.3f}\nRMSE = {rmse_val:.3f}')
ideal_line = mlines.Line2D([], [], color='gray', linestyle='--', linewidth=2, 
                           label='1:1 Perfect Prediction')

plt.legend(handles=[reg_line, ideal_line], frameon=True, loc='upper left', fontsize=10)

# 7. Formatting
plt.title(f"CoexpressDeconvolve Predictions vs. Ground Truth\n(Optimal Calibration Factor = 0.7040)", fontsize=14, fontweight='bold')
plt.xlabel("Known Cells per Spot (Synthetic Ground Truth)", fontsize=12)
plt.ylabel("Predicted Cells per Spot", fontsize=12)

# Clip at 0, since negative cells are physically impossible
plt.ylim(bottom=0)

sns.despine()
plt.tight_layout()

# Save high-resolution version for the manuscript
plt.savefig("Plot_4_Optimization_Result_600dpi.png", dpi=600, bbox_inches='tight')
plt.show()
```

```python
from scipy.stats import pearsonr
from sklearn.metrics import mean_squared_error

# 1. Define the Global UMI Baseline (Derived from Ground Truth for the benchmark)
mean_umi_per_cell = np.mean(total_umi[valid_mask] / true_cells[valid_mask])

def calculate_hybrid_rmse(alpha):
    """
    alpha = weight given to UMI. (1 - alpha) = weight given to HK.
    """
    # A. Calculate HK-based cell prediction (using our optimized factor 0.7040)
    adjusted_log_mean = standard_anchor_log_mean + np.log(0.7040)
    hk_signal_ref = np.expm1(adjusted_log_mean)
    pred_hk = spot_signal_linear / hk_signal_ref
    
    # B. Calculate UMI-based cell prediction
    pred_umi = total_umi / mean_umi_per_cell
    
    # C. Hybrid Geometric Mean
    # Using geometric mean prevents either metric from dragging the other too far
    pred_hybrid = (pred_umi ** alpha) * (pred_hk ** (1 - alpha))
    
    # Mask invalid spots
    pred_hybrid[~valid_mask] = 0
    
    # Calculate RMSE
    rmse = np.sqrt(mean_squared_error(true_cells[valid_mask], pred_hybrid[valid_mask]))
    return rmse, pred_hybrid

# 2. Test a balanced Hybrid (alpha = 0.5)
alpha_test = 0.5
hybrid_rmse, hybrid_preds = calculate_hybrid_rmse(alpha_test)

# 3. Calculate new Pearson R!
r_hybrid, _ = pearsonr(true_cells[valid_mask], hybrid_preds[valid_mask])

print(f"--- HYBRID ANCHOR RESULTS (50% UMI / 50% HK) ---")
print(f"Pearson R: {r_hybrid:.3f}   <-- LOOK AT THIS NUMBER")
print(f"RMSE:      {hybrid_rmse:.3f}")

# 4. For fun, let the optimizer find the mathematically perfect blend
from scipy.optimize import minimize_scalar

res = minimize_scalar(lambda a: calculate_hybrid_rmse(a)[0], bounds=(0.0, 1.0), method='bounded')
optimal_alpha = res.x
opt_rmse, opt_preds = calculate_hybrid_rmse(optimal_alpha)
r_opt, _ = pearsonr(true_cells[valid_mask], opt_preds[valid_mask])

print(f"\n--- OPTIMIZED HYBRID ANCHOR ---")
print(f"Best UMI Weight (Alpha): {optimal_alpha*100:.1f}%")
print(f"Best HK Weight:          {(1-optimal_alpha)*100:.1f}%")
print(f"Optimized Pearson R:     {r_opt:.3f}")
print(f"Optimized RMSE:          {opt_rmse:.3f}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr, spearmanr
from sklearn.metrics import mean_squared_error
import matplotlib.lines as mlines

# Set publication-ready style
sns.set_style("whitegrid")
sns.set_context("paper", font_scale=1.2)

# --- 1. Define the CCC Function ---
def calculate_ccc(x, y):
    cor = np.corrcoef(x, y)[0][1]
    mean_x, mean_y = np.mean(x), np.mean(y)
    var_x, var_y = np.var(x), np.var(y)
    sd_x, sd_y = np.std(x), np.std(y)
    
    numerator = 2 * cor * sd_x * sd_y
    denominator = var_x + var_y + (mean_x - mean_y)**2
    return numerator / denominator

# --- 2. Calculate the 0.5 Hybrid Blend Predictions ---
# Base HK cell estimation (using the optimized 0.7040 factor)
hk_ref = np.expm1(standard_anchor_log_mean + np.log(0.7040))
hk_cells = spot_signal_linear / hk_ref

# Calculate Global UMI baseline (just like in the codeconv.py patch)
if np.sum(hk_cells[valid_mask]) > 0:
    global_umi_per_cell = np.sum(total_umi[valid_mask]) / np.sum(hk_cells[valid_mask])
else:
    global_umi_per_cell = 1.0 

umi_cells = total_umi / global_umi_per_cell

# Apply the 0.5 Alpha Blend (Geometric Mean)
alpha = 0.6
pred_hybrid = (umi_cells ** alpha) * (hk_cells ** (1.0 - alpha))
pred_hybrid[~valid_mask] = 0

# Round to discrete cell counts
final_predicted_cells = np.round(pred_hybrid).astype(int)

# --- 3. Calculate Final Metrics ---
x_true = true_cells[valid_mask]
y_pred = final_predicted_cells[valid_mask]

r_val, _ = pearsonr(x_true, y_pred)
rho_val, _ = spearmanr(x_true, y_pred)
ccc_val = calculate_ccc(x_true, y_pred)
rmse_val = np.sqrt(mean_squared_error(x_true, y_pred))

print(f"--- 0.6 HYBRID BLEND RESULTS ---")
print(f"Pearson r: {r_val:.3f}")
print(f"Spearman p:{rho_val:.3f}")
print(f"CCC:       {ccc_val:.3f}")
print(f"RMSE:      {rmse_val:.3f} cells\n")

# --- 4. Prepare Data for Plotting ---
plot_df = pd.DataFrame({
    'True_Cells': x_true,
    'Predicted_Cells': y_pred
})

plt.figure(figsize=(9, 7))

# Plot Violins
ax = sns.violinplot(
    data=plot_df, 
    x='True_Cells', 
    y='Predicted_Cells', 
    inner=None, 
    palette="mako", 
    density_norm="width",
    hue='True_Cells',
    legend=False
)

x_min = plot_df['True_Cells'].min()

# Overlay Red Regression Line
sns.regplot(
    data=plot_df, 
    x=plot_df['True_Cells'] - x_min, 
    y='Predicted_Cells', 
    scatter=False, 
    color='#e74c3c', 
    line_kws={'linewidth': 2.5},
    ax=ax
)

# Overlay 1:1 Perfect Prediction Line
x_vals = np.sort(plot_df['True_Cells'].unique())
plt.plot(
    x_vals - x_min, x_vals, 
    color='gray', linestyle='--', linewidth=2, zorder=4
)

# Custom legend with all 4 metrics
stats_text = (f'Regression Trend\n'
              f'Pearson $r$ = {r_val:.3f}\n'
              f'Spearman $\\rho$ = {rho_val:.3f}\n'
              f'CCC = {ccc_val:.3f}\n'
              f'RMSE = {rmse_val:.3f}')

reg_line = mlines.Line2D([], [], color='#e74c3c', linewidth=2.5, label=stats_text)
ideal_line = mlines.Line2D([], [], color='gray', linestyle='--', linewidth=2, label='1:1 Perfect Prediction')

plt.legend(handles=[reg_line, ideal_line], frameon=True, loc='upper left', fontsize=11, 
           facecolor='white', framealpha=0.9)

# Formatting
plt.title(f"CoexpressDeconvolve vs. Ground Truth", 
          fontsize=15, fontweight='bold', pad=15)
plt.xlabel("Known Cells per Spot (Synthetic Ground Truth)", fontsize=13)
plt.ylabel("Predicted Cells per Spot", fontsize=13)

# Ensure y-axis doesn't go below 0
plt.ylim(bottom=-0)

sns.despine()
plt.tight_layout()

# Save high-resolution version
plt.savefig("Hybrid_0.6_Violin_600dpi.png", dpi=600, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
from sklearn.metrics import mean_squared_error

# --- 1. Define the Metrics (Exactly as in the Direct Plot) ---
def calculate_ccc(x, y):
    cor = np.corrcoef(x, y)[0][1]
    mean_x, mean_y = np.mean(x), np.mean(y)
    var_x, var_y = np.var(x), np.var(y)
    sd_x, sd_y = np.std(x), np.std(y)
    numerator = 2 * cor * sd_x * sd_y
    denominator = var_x + var_y + (mean_x - mean_y)**2
    return numerator / denominator

# --- 2. Setup the Discrete Sweep Logic ---
# Use the optimized HK factor we found earlier
hk_ref = np.expm1(standard_anchor_log_mean + np.log(0.7040))
hk_cells_base = spot_signal_linear / hk_ref

# Use the standard total_umi (no bias)
# Calculate Global UMI baseline once (anchored to the HK estimate)
global_umi_per_cell = np.sum(total_umi[valid_mask]) / np.sum(hk_cells_base[valid_mask])
umi_cells_base = total_umi / global_umi_per_cell

alphas = np.linspace(0.0, 1.0, 101)
results = []

for a in alphas:
    # Exact Pipeline Logic: Hybrid -> Mask -> Round
    pred_raw = (umi_cells_base ** a) * (hk_cells_base ** (1.0 - a))
    pred_raw[~valid_mask] = 0
    pred_discrete = np.round(pred_raw).astype(int)
    
    # Calculate metrics against ground truth
    x = true_cells[valid_mask]
    y = pred_discrete[valid_mask]
    
    results.append({
        'alpha': a,
        'pearson': pearsonr(x, y)[0],
        'ccc': calculate_ccc(x, y),
        'rmse': np.sqrt(mean_squared_error(x, y))
    })

res_df = pd.DataFrame(results)

# --- 3. Visualize the "Analytical Landscape" ---
sns.set_style("whitegrid")
fig, ax1 = plt.subplots(figsize=(10, 6))

# Plot Pearson and CCC
color_p = '#2980b9'
color_c = '#27ae60'
ax1.set_xlabel('Alpha (UMI Weight)', fontsize=12)
ax1.set_ylabel('Correlation Metrics', fontsize=12)
line1 = ax1.plot(res_df['alpha'], res_df['pearson'], color=color_p, lw=3, label='Pearson r')
line2 = ax1.plot(res_df['alpha'], res_df['ccc'], color=color_c, lw=3, linestyle=':', label='CCC')
ax1.tick_params(axis='y')

# Create second axis for RMSE
ax2 = ax1.twinx()
color_r = '#c0392b'
ax2.set_ylabel('RMSE (Cells)', color=color_r, fontsize=12)
line3 = ax2.plot(res_df['alpha'], res_df['rmse'], color=color_r, lw=3, linestyle='--', label='RMSE')
ax2.tick_params(axis='y', labelcolor=color_r)

# Mark the 0.5 Blend
plt.axvline(x=0.5, color='black', alpha=0.3, linestyle='--')
# Annotate the exact values at 0.5 to prove they match your direct plot
val_05 = res_df.iloc[50]
ax1.annotate(f"At Alpha 0.5:\n$r$={val_05['pearson']:.3f}\nRMSE={val_05['rmse']:.3f}", 
             xy=(0.5, val_05['pearson']), xytext=(0.55, val_05['pearson']-0.05),
             arrowprops=dict(arrowstyle='->'), fontsize=10, bbox=dict(facecolor='white', alpha=0.8))

plt.title("Discrete Optimization Landscape (Exact Pipeline Logic)", fontsize=14, fontweight='bold')
lines = line1 + line2 + line3
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='lower left', frameon=True)

plt.tight_layout()
plt.show()
```
