```python
import h5py
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr, spearmanr

# 1. Load Ground Truth and Captured UMI Data
# The ground_truth_counts.csv contains the simulated True_Cells [cite: 860]
gt = pd.read_csv('ground_truth_counts.csv')
gt = gt.rename(columns={'Spot': 'barcode', 'True_Cells': 'n_cells'})

# Extracting raw captured signal from the Visium matrix [cite: 782, 783]
h5_path = 'filtered_feature_bc_matrix.h5' 
with h5py.File(h5_path, 'r') as f:
    barcodes = [b.decode('utf-8') for b in f['matrix']['barcodes'][:]]
    data = f['matrix']['data'][:]
    indptr = f['matrix']['indptr'][:]
    umi_counts = [np.sum(data[indptr[i]:indptr[i+1]]) for i in range(len(barcodes))]

umi_df = pd.DataFrame({'barcode': barcodes, 'total_umi': umi_counts})
df = pd.merge(gt, umi_df, on='barcode')

# 2. Statistics for Annotation
p_val, _ = pearsonr(df['n_cells'], df['total_umi'])
s_val, _ = spearmanr(df['n_cells'], df['total_umi'])

# 3. Create High-Resolution Visualization (600 DPI)
plt.figure(figsize=(10, 8.5))

# Plot Violins: inner=None removes individual data points for a cleaner distribution view
sns.violinplot(
    data=df, 
    x='n_cells', 
    y='total_umi', 
    inner=None, 
    palette="viridis",
    scale="width"
)

# Add Red Correlation (Regression) Line
# Note: Categorical violin plots start at index 0. 
# We subtract the minimum cell count to align the numeric regression line with the plot bins.
sns.regplot(
    data=df, 
    x=df['n_cells'] - df['n_cells'].min(), 
    y='total_umi', 
    scatter=False, 
    color='red', 
    line_kws={'linewidth': 2.5, 'label': 'Regression Line'}
)

# Formatting
plt.title(f"Transcriptional Yield Scaling vs. Cellular Density\nPearson R = {p_val:.3f} | Spearman R = {s_val:.3f}")
plt.xlabel("Number of Cells (Synthetic Ground Truth)")
plt.ylabel("Total UMI Captured per Spot")
plt.legend()

sns.despine()
plt.grid(axis='y', alpha=0.15)

# Save at 600 DPI for publication quality
plt.savefig("umi_violin_with_regression_600dpi.png", dpi=600, bbox_inches='tight')
plt.show()
```
