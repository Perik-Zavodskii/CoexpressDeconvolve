```r
if (!require("BiocManager", quietly = TRUE))
    install.packages("BiocManager")

BiocManager::install("zellkonverter")
```

```r
library(Seurat)
library(zellkonverter)
library(SingleCellExperiment)
```

```r
# 1. Load Data
sc_obj <- readRDS("Seurat_Clusters.rds")
hvgs <- read.csv("fixed_hvgs.csv", header=FALSE)[,1]
```

```r
# 2. Subset to Curated HVGs
sc_obj <- subset(sc_obj, features = hvgs)

# 3. CRITICAL FOR SEURAT V5: Join layers before conversion
sc_obj <- JoinLayers(sc_obj)

# 4. Convert to SingleCellExperiment, then to .h5ad
sce <- as.SingleCellExperiment(sc_obj)
writeH5AD(sce, file = "sc_reference_hvgs.h5ad")

print("Prep complete: sc_reference_hvgs.h5ad saved.")
```
