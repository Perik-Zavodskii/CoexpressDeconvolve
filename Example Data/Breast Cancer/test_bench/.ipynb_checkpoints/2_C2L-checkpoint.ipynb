```python
import scanpy as sc
import pandas as pd
import numpy as np
import cell2location
from cell2location.models import RegressionModel, Cell2location
import warnings
import scipy.sparse as sp
import h5py
from scipy.sparse import csr_matrix

warnings.filterwarnings('ignore')
```

```python
# File Paths

sc_path = "sc_reference_full.h5ad" 
hvg_path_synth = "fixed_hvgs.csv"
hvg_path_real = "fixed_hvgs_real.csv"

synth_sp_path = "/Users/romanperikzavodskii/Desktop/4. Ery/CoDeconv/Breast Cancer/test_bench/Synthetic Visium/filtered_feature_bc_matrix.h5"
real_sp_path = '/Users/romanperikzavodskii/Desktop/4. Ery/CoDeconv/Breast Cancer/Visium/Breast Cancer/filtered_feature_bc_matrix.h5'

# Load Single-Cell Reference

print("Loading Single-Cell Reference...")
ad_sc = sc.read_h5ad(sc_path)
ad_sc.obs['CellType'] = ad_sc.obs['CellType'].astype('category')

# Load Curated HVGs for both pipelines
hvgs_synth = pd.read_csv(hvg_path_synth, header=None)[0].tolist()
hvgs_real = pd.read_csv(hvg_path_real, header=None)[0].tolist()

print(f"Loaded {ad_sc.n_obs} cells and {ad_sc.n_vars} total genes.")


# Calculate Average GEX per Cell Type (For Reconstructing Spatial GEX)

print("\nCalculating average Whole Transcriptome expression per Cell Type...")
cell_types = ad_sc.obs['CellType'].unique()
genes = ad_sc.var_names

avg_array = np.zeros((len(cell_types), len(genes)))

for i, ct in enumerate(cell_types):
    cells_of_type = ad_sc[ad_sc.obs['CellType'] == ct]
    if sp.issparse(cells_of_type.X):
        avg_array[i, :] = np.array(cells_of_type.X.mean(axis=0))
    else:
        avg_array[i, :] = cells_of_type.X.mean(axis=0)

sc_avg_expr = pd.DataFrame(avg_array, index=cell_types, columns=genes)
print(f"Average expression matrix calculated. Shape: {sc_avg_expr.shape}")
```

```python
# Helper Functions

def robust_read_10x_h5(filename):
    """Manually loads 10x H5 file to avoid empty feature column errors."""
    with h5py.File(filename, 'r') as f:
        g = f['matrix']
        X = csr_matrix((g['data'][:], g['indices'][:], g['indptr'][:]), 
                       shape=g['shape'][:][::-1])
        obs = pd.DataFrame(index=g['barcodes'][:].astype(str))
        features = g['features']
        var = pd.DataFrame({
            'gene_ids': features['id'][:].astype(str),
            'feature_types': features['feature_type'][:].astype(str)
        }, index=features['name'][:].astype(str))
        adata = sc.AnnData(X=X, obs=obs, var=var)
        adata.var_names_make_unique()
        return adata

def get_c2l_signatures(ad_sc, hvgs):
    print("Subsetting Single-Cell reference to HVGs for training...")
    ad_sc_train = ad_sc[:, ad_sc.var_names.isin(hvgs)].copy()
    RegressionModel.setup_anndata(adata=ad_sc_train, labels_key='CellType')

    print("Training Reference Signatures (This may take a while)...")
    sc_model = RegressionModel(ad_sc_train)
    sc_model.train(max_epochs=250, batch_size=1000, train_size=1, lr=0.002)

    ad_sc_train = sc_model.export_posterior(ad_sc_train, sample_kwargs={'num_samples': 1000, 'batch_size': 1000})
    inf_aver = ad_sc_train.varm['means_per_cluster_mu_fg']
    inf_aver.columns = ad_sc_train.uns['mod']['factor_names']
    print("Reference signatures successfully learned!")
    return inf_aver

def run_c2l_pipeline(sp_path, signatures, sc_avg_expr, prefix="c2l_synth", is_synth=True):
    print(f"\nStarting Spatial Pipeline for: {prefix}")
    
    # 1. Load Spatial Data
    print("Loading Spatial Data...")
    try:
        ad_sp = sc.read_10x_h5(sp_path)
    except ValueError:
        print("Standard read failed. Using robust manual loader...")
        ad_sp = robust_read_10x_h5(sp_path)
    
    ad_sp.var_names_make_unique()
    
    # 2. Subset to Shared HVGs
    shared_genes = list(set(signatures.index) & set(ad_sp.var_names))
    ad_sp_train = ad_sp[:, shared_genes].copy()
    signatures_train = signatures.loc[shared_genes]
    
    # 3. Setup and Train Spatial Model
    Cell2location.setup_anndata(adata=ad_sp_train)
    mod = Cell2location(ad_sp_train, cell_state_df=signatures_train, N_cells_per_location=5)
    
    print("Training Spatial Model (CPU)...")
    mod.train(max_epochs=15000, batch_size=None, train_size=1, accelerator="cpu")
    
    print("Exporting posterior...")
    ad_sp_train = mod.export_posterior(ad_sp_train, sample_kwargs={'num_samples': 1000, 'batch_size': ad_sp_train.n_obs})
    
    # 4. Extract Abundance (nCell Counts)
    abundance_df = ad_sp_train.obsm['q05_cell_abundance_w_sf']
    abundance_df.columns = [c.replace('q05cell_abundance_w_sf_', '') for c in abundance_df.columns]
    
    # 5. Save results 
    # Notice we save proportions for comparison, but we also save counts (which is native to c2l)
    abundance_df.to_csv(f"{prefix}_counts.csv")
    if is_synth:
        props_df = abundance_df.div(abundance_df.sum(axis=1), axis=0).fillna(0)
        props_df.to_csv(f"{prefix}_props.csv")
        print(f"Saved synthetic metrics (counts & proportions).")
        
    # 6. Reconstruct Whole Transcriptome (WT) GEX
    print("Reconstructing WT pseudobulk...")
    wt_gex_df = abundance_df.dot(sc_avg_expr.loc[abundance_df.columns])
    wt_gex_df.to_csv(f"{prefix}_wt_gex.csv")
    print(f"Saved: {prefix}_wt_gex.csv")
    
    return ad_sp_train
```

```python
# 1. Run on Synthetic Data

print("\n=== RUNNING SYNTHETIC PIPELINE ===")
# Generate signatures specifically using the synthetic HVGs
inf_aver_synth = get_c2l_signatures(ad_sc, hvgs_synth)

ad_sp_synth = run_c2l_pipeline(
    sp_path=synth_sp_path, 
    signatures=inf_aver_synth, 
    sc_avg_expr=sc_avg_expr, 
    prefix="c2l_synth", 
    is_synth=True
)
```

```python
# 2. Run on Real Data

print("\n=== RUNNING REAL PIPELINE ===")
# Generate signatures specifically using the real HVGs
inf_aver_real = get_c2l_signatures(ad_sc, hvgs_real)

ad_sp_real = run_c2l_pipeline(
    sp_path=real_sp_path, 
    signatures=inf_aver_real, 
    sc_avg_expr=sc_avg_expr, 
    prefix="c2l_real", 
    is_synth=False
)

print("\ncell2location benchmark runs complete!")
```
